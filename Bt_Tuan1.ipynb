Bài 3

So sánh 2 mảng 1 chiều, 2 chiều bằng các hàm greater, greater_equal, less, less_equal.

```python
import numpy as np

# So sánh mảng 1 chiều
a = np.array([1, 5, 8, 10])
b = np.array([2, 5, 6, 12])

print("Mảng 1 chiều a:", a)
print("Mảng 1 chiều b:", b)

print("a > b:", np.greater(a, b))
print("a >= b:", np.greater_equal(a, b))
print("a < b:", np.less(a, b))
print("a <= b:", np.less_equal(a, b))

print("\n----------------------\n")

# So sánh mảng 2 chiều
x = np.array([[1, 5], [8, 10]])
y = np.array([[2, 5], [6, 12]])

print("Mảng 2 chiều x:")
print(x)

print("Mảng 2 chiều y:")
print(y)

print("x > y:")
print(np.greater(x, y))

print("x >= y:")
print(np.greater_equal(x, y))

print("x < y:")
print(np.less(x, y))

print("x <= y:")
print(np.less_equal(x, y))
```

Bài 17

Viết chương trình Pandas sắp xếp theo cột của DataFrame 4 x 3 ngẫu nhiên.

```python
import pandas as pd
import numpy as np

# Tạo DataFrame 4 x 3 ngẫu nhiên
df = pd.DataFrame(
    np.random.randint(1, 100, size=(4, 3)),
    columns=['A', 'B', 'C']
)

print("DataFrame ban đầu:")
print(df)

# Sắp xếp theo cột A
df_sort_A = df.sort_values(by='A')

print("\nDataFrame sau khi sắp xếp theo cột A:")
print(df_sort_A)

# Sắp xếp theo cột B
df_sort_B = df.sort_values(by='B')

print("\nDataFrame sau khi sắp xếp theo cột B:")
print(df_sort_B)

# Sắp xếp theo cột C
df_sort_C = df.sort_values(by='C')

print("\nDataFrame sau khi sắp xếp theo cột C:")
print(df_sort_C)
```

Bài 19

```python
import pandas as pd

sr1 = pd.Series([1, 2, 3, 4, 5])
sr2 = pd.Series([2, 4, 6, 8, 10])

print("sr1:")
print(sr1)

print("\nsr2:")
print(sr2)

result = sr1[~sr1.isin(sr2)]

print("\nItems of sr1 not present in sr2:")
print(result)
```

Bài 21

Viết chương trình Pandas thêm Series y vào Series x theo chiều dọc và chiều ngang.

```python
import pandas as pd

# Tạo Series x
x = pd.Series(range(10))

# Tạo Series y
y = pd.Series(list('pqrstuvwxy'))

print("Series x")
print(x)

print("\nSeries y")
print(y)

# Thêm series y vào series x theo chiều dọc
vertical = pd.concat([x, y], axis=0)

print("\nThêm series y vào series x theo chiều dọc vertical:")
print(vertical)

# Thêm series y vào series x theo chiều ngang
horizontal = pd.concat([x, y], axis=1)

print("\nThêm series y vào series x theo chiều ngang horizontal:")
print(horizontal)
```
